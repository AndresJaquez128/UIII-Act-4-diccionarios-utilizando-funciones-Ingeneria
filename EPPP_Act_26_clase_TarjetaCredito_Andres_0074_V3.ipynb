```python
Diccionario_Ingenería = []


def agregar_Ingenería():

    nombre = input("Ingrese el nombre: ")
    carrera = input("Ingrese la carrera: ")
    semestre = input("Ingrese el semestre: ")
    promedio = input("Ingrese el promedio: ")

    datos = {
        "Nombre": nombre,
        "Carrera": carrera,
        "Semestre": semestre,
        "Promedio": promedio
    }

    Diccionario_Ingenería.append(datos)

    print("\nDATOS AGREGADOS CORRECTAMENTE\n")

    Listar_Ingenería()


def Listar_Ingenería():

    print("\n==================== LISTA DE INGENIERÍA ====================")

    print("--------------------------------------------------------------------------")

    print("{:<20} {:<20} {:<15} {:<15}".format(
        "NOMBRE", "CARRERA", "SEMESTRE", "PROMEDIO"
    ))

    print("--------------------------------------------------------------------------")

    for datos in Diccionario_Ingenería:

        for key, value in datos.items():

            print("{:<20}".format(value), end=" ")

        print()

    print("--------------------------------------------------------------------------\n")


def buscar_Ingenería():

    buscar = input("Ingrese el nombre a buscar: ")

    print("\n================== RESULTADO DE BÚSQUEDA ==================")

    print("--------------------------------------------------------------------------")

    print("{:<20} {:<20} {:<15} {:<15}".format(
        "NOMBRE", "CARRERA", "SEMESTRE", "PROMEDIO"
    ))

    print("--------------------------------------------------------------------------")

    for datos in Diccionario_Ingenería:

        if datos["Nombre"] == buscar:

            for key, value in datos.items():

                print("{:<20}".format(value), end=" ")

            print()

    print("--------------------------------------------------------------------------\n")


def borrar_Ingenería():

    borrar = input("Ingrese el nombre a borrar: ")

    for datos in Diccionario_Ingenería:

        if datos["Nombre"] == borrar:

            Diccionario_Ingenería.remove(datos)

            print("\nDATOS BORRADOS CORRECTAMENTE\n")

    Listar_Ingenería()


def actualizar_Ingenería():

    actualizar = input("Ingrese el nombre a actualizar: ")

    for datos in Diccionario_Ingenería:

        if datos["Nombre"] == actualizar:

            datos["Carrera"] = input("Nueva carrera: ")
            datos["Semestre"] = input("Nuevo semestre: ")
            datos["Promedio"] = input("Nuevo promedio: ")

            print("\nDATOS ACTUALIZADOS CORRECTAMENTE\n")

    Listar_Ingenería()


while True:

    print("================ MENÚ ================")
    print("1. Agregar Ingeniería en un diccionario")
    print("2. Listar diccionarios de Ingeniería")
    print("3. Buscar Ingeniería en diccionario")
    print("4. Borrar Ingeniería en diccionario")
    print("5. Actualizar Ingeniería en diccionario")
    print("6. Salir")
    print("======================================")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":

        agregar_Ingenería()

    elif opcion == "2":

        Listar_Ingenería()

    elif opcion == "3":

        buscar_Ingenería()

    elif opcion == "4":

        borrar_Ingenería()

    elif opcion == "5":

        actualizar_Ingenería()

    elif opcion == "6":

        print("\nPROGRAMA FINALIZADO")
        break

    else:

        print("\nOPCIÓN INCORRECTA\n")
print("Software desarrollado por Jaquez Andres NL = 27")
```
